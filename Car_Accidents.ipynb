Normalizacion de datos de columna city

```python
city_normalized = (
    car_accidents['City']
    .dropna()
    .str.strip()
    .str.lower()
)

city_variants = (
    pd.DataFrame({
        'original': car_accidents.loc[city_normalized.index, 'City'],
        'normalized': city_normalized
    })
    .groupby('normalized')['original']
    .nunique()
    .sort_values(ascending=False)
)

city_variants[city_variants > 1]
```

```python
city_normalized = car_accidents['City'].str.strip().str.casefold()

city_variants = (
    pd.DataFrame({
        'City_original': car_accidents['City'],
        'City_normalized': city_normalized
    })
    .dropna()
    .groupby('City_normalized')['City_original']
    .unique()
)

city_variants = city_variants[city_variants.apply(len) > 1]

city_variants
```

```python
city_counts = (
    car_accidents
    .dropna(subset=['City'])
    .assign(
        City_normalized=lambda x: x['City'].str.strip().str.casefold()
    )
    .groupby(['City_normalized', 'City'], observed=True)
    .size()
)
```

```python
for city in city_variants.index:
    print(f"\n{city}")
    
    mask = city_normalized == city
    
    print(
        car_accidents.loc[mask, 'City']
        .value_counts()
    )
```

```python
city_corrections = {}

for normalized, variants in city_variants.items():
    counts = (
        car_accidents.loc[
            city_normalized.eq(normalized),
            'City'
        ]
        .value_counts()
    )
    
    canonical = counts.index[0]
    
    for variant in variants:
        if variant != canonical:
            city_corrections[variant] = canonical
```

```python
for original, corrected in city_corrections.items():
    print(f"{original} → {corrected}")
```

```python
car_accidents['City'] = car_accidents['City'].astype('string')
car_accidents['City'] = car_accidents['City'].replace(city_corrections)
```

### Analisis de infraestructura

```python
infrastructure_cols = [
    'Amenity',
    'Bump',
    'Crossing',
    'Give_Way',
    'Junction',
    'No_Exit',
    'Railway',
    'Roundabout',
    'Station',
    'Stop',
    'Traffic_Calming',
    'Traffic_Signal',
    'Turning_Loop'
]

infrastructure_summary = accidents_severity[infrastructure_cols].sum().to_frame(name='Yes')

infrastructure_summary['No'] = (
    accidents_severity[infrastructure_cols].notna().sum()
    - infrastructure_summary['Yes']
)

infrastructure_summary
```

```python
infrastructure_severity = pd.DataFrame()

for col in infrastructure_cols:
    table = pd.crosstab(
        accidents_severity[col],
        accidents_severity['Severity']
    )
    
    infrastructure_severity[col] = table.stack()

infrastructure_severity
```
